2. It violates the rule: "Column names should be values, not variable names." Here, the subjects (Math, Science, English) are stored as column headers instead of values under a single Subject variable/column.


```python
import pandas as pd

# 1. Create the initial untidy DataFrame
data1 =  {
    'Student': ['Ali', 'Sara', 'John'],
    'Math': [75, 88, 65],
    'Science': [80, 91, 70],
    'English': [72, 85, 68]
}
df1 = pd.DataFrame(data1)

# 2. Reshape into tidy format using pd.melt()
df1_tidy = df1.melt(
    id_vars=['Student'],
    var_name='Subject',
    value_name='Score'
)

#Display the tidy DataFrame
print(df1_tidy)
```

Question 2: Sales Data

1. The column headers (Q1_Sales, Q2_Sales, Q3_Sales, Q4_Sales) combine variable values (the quarter time period) into column names. In tidy data, values like Q1, Q2, etc., belong in their own Quarter column.

```python
import pandas as pd

# 1. Create the initial untidy DataFrame
data2 = {
    'Country': ['Malaysia', 'Thailand', 'Indonesia'],
    'Q1_Sales': [12000, 10000, 14000],
    'Q2_Sales': [15000, 11000, 12500],
    'Q3_Sales': [14000, 13000, 15000],
    'Q4_Sales': [18000, 16000, 17000]
}
df2 = pd.DataFrame(data2)

# 2. Reshape into tidy format using pd.melt()
df2_tidy = df2.melt(
    id_vars=['Country'],
    var_name='Quarter',
    value_name='Sales'
)

# 3. Clean up the 'Quarter' column values (e.g., convert 'Q1_Sales' -> 'Q1')
df2_tidy['Quarter'] = df2_tidy['Quarter'].str.replace('_Sales', '')

# Display the result
print(df2_tidy)
```

```python
# Question 3: Asking Questions from Data

import pandas as pd
import matplotlib.pyplot as plt

# Dataset
data = {
    "Student": ["A", "B", "C", "D", "E", "F"],
    "Study_Hours": [2, 5, 3, 7, 4, 8],
    "Score": [55, 72, 60, 85, 68, 90]
}
df = pd.DataFrame(data)

# 1. Average study time
avg_hours = df["Study_Hours"].mean()
print(f"1. Average study time: {avg_hours:.2f} hours")

# 2. Average score
avg_score = df["Score"].mean()
print(f"2. Average score: {avg_score:.2f}")

# 3. Student with the highest score
top_student = df.loc[df["Score"].idxmax(), "Student"]
print(f"3. Highest score student: {top_student}")

# 4. Student who studied the most
most_studied = df.loc[df["Study_Hours"].idxmax(), "Student"]
print(f"4. Studied the most: {most_studied}")

# 5. Correlation check (optional helper)
corr = df["Study_Hours"].corr(df["Score"])
print(f"Correlation: {corr:.4f}")

# 6. Scatter plot of Study_Hours vs Score
plt.figure(figsize=(6, 4))
plt.scatter(df["Study_Hours"], df["Score"], color="blue")
plt.title("Study Hours vs Score")
plt.xlabel("Study Hours")
plt.ylabel("Score")
plt.grid(True)
plt.show()
```
